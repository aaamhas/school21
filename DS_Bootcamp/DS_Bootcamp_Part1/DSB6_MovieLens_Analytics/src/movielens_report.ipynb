# MovieLens

Какие фильмы попадают в базу, как зрители ставят оценки, какие фильмы становятся лидерами, а какие вызывают споры? В этом отчёте мы ищем ответы на эти вопросы на данных MovieLens.

**Данные.** MovieLens latest-small: https://grouplens.org/datasets/movielens/latest/

Чтобы отчёт запустился, распакуйте `movies.csv`, `ratings.csv`, `tags.csv` и `links.csv` в папку `datasets/` рядом с папкой `src/`.

Анализируются первые 1000 строк каждого файла.

```python
%timeit -n 1 -r 1 from movielens_analysis import Movies, Ratings
from movielens_analysis import Movies, Ratings, Tags, Links
```

```python
%timeit -n 1 -r 1 movies = Movies("../datasets/movies.csv")
%timeit -n 1 -r 1 ratings = Ratings("../datasets/ratings.csv")
movies = Movies("../datasets/movies.csv")
ratings = Ratings("../datasets/ratings.csv")
%timeit -n 1 -r 1 ratings_movies = ratings.Movies(ratings.data)
%timeit -n 1 -r 1 ratings_users = ratings.Users(ratings.data)
%timeit -n 1 -r 1 tags = Tags("../datasets/tags.csv")
%timeit -n 1 -r 1 links = Links("../datasets/links.csv")
ratings_movies = ratings.Movies(ratings.data)
ratings_users = ratings.Users(ratings.data)
tags = Tags("../datasets/tags.csv")
links = Links("../datasets/links.csv")
```

```python
%timeit tags.read_csv()
tags.read_csv()[:5]
```

## Вопрос 1. Кино каких лет собрано в базе?
Год выпуска достаём прямо из названия фильма, например «Toy Story (1995)».

```python
%timeit movies.dist_by_release()
movies.dist_by_release()
```

Больше всего фильмов вышло в 1995 году: 224

## Вопрос 2. Какие жанры самые частые?

```python
%timeit movies.dist_by_genres()
movies.dist_by_genres()
```

Лидируют драмы и комедии, а редкие жанры вроде IMAX встречаются всего несколько раз.

## Вопрос 3. У каких фильмов больше всего жанров?

```python
%timeit movies.most_genres(10)
movies.most_genres(10)
```

Рекорд — 6 жанров, и его делят сразу 8 фильмов, среди них Lion King (1994), Super Mario Bros. (1993) и Space Jam (1996).

## Вопрос 4. Когда ставились оценки?
Год оценки достаём из timestamp — это время, когда зритель нажал на звёздочку, а не год выхода фильма.

```python
%timeit ratings_movies.dist_by_year()
ratings_movies.dist_by_year()
```

Оценки распределены по 1996-2015 годам. Больше всего оценок поставлено в 1996 году (358)

## Вопрос 5. Какие оценки ставят чаще всего?

```python
%timeit ratings_movies.dist_by_rating()
ratings_movies.dist_by_rating()
```

Самая частая оценка — 4.0 (292 оценки), затем идет 5.0 (267 оценок)

## Вопрос 6. Какие фильмы оценивают чаще всего?

```python
%timeit -n 1 -r 1 ratings_movies.top_by_num_of_ratings(10)
ratings_movies.top_by_num_of_ratings(10)
```

Больше 4 оценок нет ни у одного фильма: в первых 1000 строках всего 7 зрителей.
Поэтому оценки 5.0 в следующем вопросе — это фильмы, которые оценили 1–2 человека.

## Вопрос 7. Какие фильмы оценены выше всего?
Сравним два способа: среднее и медиану.

```python
%timeit -n 1 -r 1 ratings_movies.top_by_ratings(10, 'average')
ratings_movies.top_by_ratings(10, 'average')
```

```python
%timeit -n 1 -r 1 ratings_movies.top_by_ratings(10, 'median')
ratings_movies.top_by_ratings(10, 'median')
```

В топе много фильмов с оценкой 5.0 — скорее всего, у них всего одна-две оценки.
Поэтому «лучший фильм» надо смотреть вместе с числом оценок из вопроса 6.

## Вопрос 8. О каких фильмах спорят?
Дисперсия оценок показывает, насколько зрители расходятся во мнениях: у спорного фильма есть и пятёрки, и единицы.

```python
%timeit -n 1 -r 1 ratings_movies.top_controversial(10)
ratings_movies.top_controversial(10)
```

Самые спорные фильмы — Bambi (1942), Rescuers, The (1977) и My Fair Lady (1964) с дисперсией 5.06

## Вопрос 9. Кто оценивает больше всех?

```python
%timeit ratings_users.ratings_dict()
ratings_users.ratings_dict()
```

В первых 1000 оценках всего 7 зрителей, и самый активный поставил 314 оценок.

```python
%timeit ratings_users.dist_users('median')
ratings_users.dist_users('median')
```

## Вопрос 11. Чьи оценки самые непредсказуемые?

## Вопрос 15. Что можно узнать о фильме с TMDB?
Для каждого фильма по tmdbId скачиваем страницу и достаём режиссёра, бюджет, сборы и длительность.
Чтобы не ходить в сеть каждый раз, результаты сохраняются в imdb_cache.json.

```python
%timeit -n 1 -r 1 links.get_imdb([1, 2, 3, 4, 5], ["Director", "Budget", "Revenue", "Runtime"])
links.get_imdb([1, 2, 3, 4, 5], ["Director", "Budget", "Revenue", "Runtime"])
```

## Вопрос 16. Какие режиссёры встречаются чаще всего?
